```python
#first and last interchange.
s=input("Enter a string:")
print("New string is:",s[-1]+s[1:-1]+s[0])
```

```python
#create a string from first 2 and last 2 characters of a string.
s=input("Enter a string:")
print("New string is:",s[:2]+s[-2:])
```

```python
#create a single string separated with space from two string by swapping the char position 1
s1=input("Enter the 1st string:")
s2=input("Enter the 2nd string:")
n1=s2[0]+s1[1:]
n2=s1[0]+s2[1:]
print("New string:",n1,n2)
```

```python
#replace()
s=input("Enter a string:")
ns=s[0]+s[1:].replace(s[0],"$")
print("New string:",ns)
```

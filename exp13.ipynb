```python
%%writefile rectangle.py
#rectangle module.
def area(l,b):
    return l*b
def perimeter(l, b):
    return 2*(l+b)
```

```python
%%writefile circle.py
#circle module.
import math
def area(r):
    return math.pi*r*r
def perimeter(r):
    return 2*math.pi*r
```

```python
#create two module and import them
import rectangle
import circle
while True:
    print("1.Rectangle\n2.Circle\n3.Exit\n")
    ch=int(input("Enter your chioce:"))
    if ch==1:
        l=int(input("Enter the length:"))
        b=int(input("Enter the breadth:"))
        print("Length:",l)
        print("Breadth:",b)
        print(f"Area of Rectangle:{rectangle.area(l,b):.2f}")
        print(f"Perimeter of Rectangle:{rectangle.perimeter(l,b):.2f}")
    elif ch==2:
        r=float(input("Enter the radius:"))
        print("Radius:",r)
        print(f"Area of circle:{circle.area(r):.2f}")
        print(f"Perimeter of circle:{circle.perimeter(r):.2f}")
    elif ch==3:
        print("Exiting program")
        break
    else:
        print("Invalid choice")
```

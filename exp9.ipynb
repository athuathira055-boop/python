```python
#add "ing" to the end of a string.if the ends with "ing",add "ly" instead
s=input("Enter a string:")
print("string is:",s)
if s[-3:]=="ing":
    print(s+"ly")
else:
    print(s+"ing")
```

```python
#To check the give string is palindrome or not
s=input("Enter a string:")
print("string is:",s)
if s==s[::-1]:
    print("The string is palindrome",s)
else:
    print("The string is not palindrome",s)
```

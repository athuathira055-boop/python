```python
#colors in color1 but not in color2
color1=set(input("Enter colors in color1:").split())
color2=set(input("Enter colors in color2:").split())
print("color1=",color1)
print("color2=",color2)
difference_color=color1.difference(color2)
print("The colors in color1 but not in color2:",difference_color)
```

```python
#color from both set1 and set2
colorset1=set(input("Enter the color in set1:").split())
colorset2=set(input("Enter the color in set2:").split())
print("set1=",colorset1)
print("set2=",colorset2)
print("The color in both sets are:",colorset1.union(colorset2))
```

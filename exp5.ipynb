```python
#get 3rd element from 1st and 3rd element from last of a tuple
tuple=(10,20,30,40,50,60)
print("3rd element is:",tuple[2])
print("3rd element from the last is:",tuple[-3])
```

```python
#average
number=input("Enter the number:")
tuples=tuple(map(int,number.split(",")))
print("tuple is:",tuples)
print("average is:",sum(tuples)/len(tuples))
```

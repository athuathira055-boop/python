```python
#sort the dictionary in ascending and descending order of their roll number
roll=input("enter roll number:").split()
name=input("enter student name:").split()
my_dict=dict(zip(roll,name))
print("ascending:",dict(sorted(my_dict.items())))
print("descending:",dict(sorted(my_dict.items())))
```

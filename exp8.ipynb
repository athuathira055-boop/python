```python
#sort the dictionary in ascending and descending order of their roll number
roll=input("enter roll number:").split()
name=input("enter student name:").split()
my_dict=dict(zip(roll,name))
print("orginal dict is:",my_dict)
print("ascending:",dict(sorted(my_dict.items())))
print("descending:",dict(sorted(my_dict.items())))
```

```python
#merge 2 dict using update()
dict_1=dict(zip(input("Enter 3 roll number:").split(),input("Enter 3 names:").split()))
dict_2=dict(zip(input("Enter 2 roll number:").split(),input("Enter 2 names:").split()))
print("dictionary1=",dict_1)
print("dictionary2=",dict_2)
dict_1.update(dict_2)
print("merge dict:",dict_1)
```

```python
#sort the dictionary in ascending and descending order of their roll number
roll=input("enter roll number:").split()
name=input("enter student name:").split()
my_dict=dict(zip(roll,name))
print("orinsl dictionary is:",my_dict)
print("ascending:",dict(sorted(my_dict.items())))
print("descending:",dict(sorted(my_dict.items(),reverse=True)))
```

```python
#merge two dictionaries using update()
dic_1=dict(zip(input("Enter 3 roll number:").split(),input("Enter 3 names:").split()))
dic_2=dict(zip(input("Enter 2 roll number:").split(),input("Enter 2 names:").split()))
print("dictionary1=",dic_1)
print("dictionary2=",dic_2)
dic_1.update(dic_2)
print("Merged dictionary:",dic_1)
```
